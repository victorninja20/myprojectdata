```python
import duckdb
```

Seleciona o banco de dados

```python
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

Vamos ver como esta o banco de dados da camada silver

```python
df =con.execute("SELECT * FROM produtos").fetchdf()
df.head(10)
```

Agora vamos criar uma dimensão na camada gold, vamos modificar o df2 e remover as colunas id_categoria e id_fornecedor

```python
df2 =df.drop(columns=['id_categoria', 'id_fornecedor'])
df2.head(10)
```

Agora vamos cria a nossa tabela na camada gold para poder fazer a insersão dos dados

repare que inserimos os nomes das tabelas em maiusculo, podemos alterar utilizando a função ALTER ou excluir toda a tabela com a função DROP, como boas praticas utilize sempre nomes em minusculo

```python
con.execute("""
    DROP TABLE IF EXISTS dim_produtos
""")
```

```python
con.execute("""
    CREATE TABLE IF NOT EXISTS dim_produtos (
            id_produto BIGINT,
            nm_produto TEXT,
            vl_produto FLOAT
            )
         
 """)
```

FAMOS FAZER UMA CONSULTA NA dim_produtos

```python
df_dim = con.execute("SELECT * FROM dim_produtos").fetchdf()
df_dim.head(10)
```

Agora vamos fazer a insersão dos dados na tabela dim_produtos como os dados vindo do dataframe df2

```python
con.execute("INSERT INTO dim_produtos SELECT * FROM df2")
con.execute("SELECT * FROM dim_produtos").fetchdf().head(10)       
```

Agora precisa fechar a conexão

```python
con.close()
```
